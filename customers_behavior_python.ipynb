```python
import pandas as pd
df=pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
#checking the information about the data file
df.info()
```

```python
#checking the statistics and numerics for all columns
df.describe(include='all')
```

```python
#checking missing values 
df.isnull().sum()
```

```python
#replacing missing values with the median value based on category's
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
#checking for missing values or updated or not
df.isnull().sum()
```

```python
#changing the column names to snake casing because it will be easy to perform column operations
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
#renaming a column
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
 df.columns
```

```python
#creating a age_group
labels=['Young-Adult','Adult','Middle-Aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
#checking the group added or not and how it organised
df[['age','age_group']].head(10)
```

```python
#creating a column to convert the frequency purchase text to numeric so it can be easy to analysis
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
#checking the updation of new columns
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
#analysing about discount and promo code applied on both or not
df[['discount_applied','promo_code_used']].head(10)
```

```python
#checking both appiled or not 
(df['discount_applied']==df['promo_code_used']).all()
```

```python
#both are used so now i will remove the promo_code_used column from the tabel
df=df.drop('promo_code_used',axis=1)
```
